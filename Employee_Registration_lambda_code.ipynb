```python
import boto3 #used to call AWS Services
```

```python
s3 = boto3.client("s3")
rekognition = boto3.client("rekognition", region_name = "Canda")
dynamodbTableName = "employee"
dynamodb = boto3.resource("dynamodb", region_name = "Canada")
employeeTable = dynamodb.Table(dynamodbTableName)
```

```python
def lambda_handler(event, context): #triggered by s3 bucket and the bucket name and the file name wll bein the event object
    print(event) # incase of errors
    bucket = event["Records"][0]["s3"]["bucket"]["name"] #bucket name
    key = event["Records"][0]["s3"]["object"]["key"] # image name of employee
    
    try: # indexing image
        response = index_employee_image(bucket, key)
        print(response) 
        if response["ResponseMetadata"]["HTTPSStatusCode"] == 200:#website 's server sends to the browser to indicate whether or not that request can be fulfilled 200 means success
            faceId = response["FaceRecords"][0]["Face"]["FaceId"] # used to identify people
            name = key.split(".")[0].split("-") # firstname_lastname.jpeg
            firstName = name[0]
            lastName = name[1]
            register_employee(faceid, first_name, last_name)
    except Exception as e:
        print(e)
        print("Error")
        raise e
        
        
```

```python
def index_employee_image(bucket, key):
    response = rekognition.index_faces(
    Image = {
        "S3Object":
        {
            "Bucket": bucket,
            "Name": key
        }
    },
    CollectionId = "employee" # make latter to save
    )
    return response
def register_employee(faceid, first_name, last_name):
    employeeTable.put_item(
    Item = {
        "rekognitionId" : faceId,
        "firstname" : first_name
        "lastName" : last_name
    }
    )
```
